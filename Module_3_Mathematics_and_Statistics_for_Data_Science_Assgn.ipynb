```python
#Matrix Addition:

matrix_a = np.array([ [2, 1, 3], [1, 0, 2], [4, 1, 1] ])
matrix_b = np.array([ [1, 2, 0], [3, 1, 4], [2, 5, 1] ])

print(matrix_a,'\n\n',matrix_b)
```

```python
#Determinant of matrix_a
det_mat_a=np.linalg.det(matrix_a)
print("Determinent of matrix A is:",det_mat_a)

det_mat_b=np.linalg.det(matrix_b)
print("Determinent of matrix B is:",det_mat_b)
```

```python
#Inverse of matrix_a
if det_mat_a!=0:
  mat_a_inverse=np.linalg.inv(matrix_a)
  print("Inverse of matrix_a is:",'\n',mat_a_inverse)

else:
  print("Matrix is singular;it has no inverse")
```

```python
#eigenvalues and eigenvectors of matrix_a
matrix_a = np.array([ [2, 1, 3],
                      [1, 0, 2],
                       [4, 1, 1] ] )
eigenvalues, eigenvectors=np.linalg.eig(matrix_a)

print("Eigen value of matrix_a is:",'\n',eigenvalues)
print("Eigen vector of matrix_b is:",'\n',eigenvectors)

```

**Task 3 – Descriptive Statistics**

Statistical Dataset

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

Activity:

Using NumPy / Pandas / statistics library:

● Compute mean, median, and mode of sales_data

● Calculate range, variance, and standard deviation

● Measure skewness and kurtosis

● Interpret whether the dataset is symmetric, positively skewed, or negatively skewed



```python

from scipy.stats import skew, kurtosis

sales_data = [120, 135, 150, 145, 160, 155, 170, 165, 180, 175,190, 200, 210, 195, 185, 175, 165, 155, 145, 135]

# Mean
mean = np.mean(sales_data)

# Median
median = np.median(sales_data)

# Mode
mode = statistics.multimode(sales_data)

# Range
data_range = np.max(sales_data) - np.min(sales_data)

# Variance
variance = np.var(sales_data)

# Standard deviation
std_dev = np.std(sales_data)

# Skewness
skewness = skew(sales_data)

# Kurtosis
kurtosis_value = kurtosis(sales_data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Range:", data_range)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
print("Skewness:", skewness)
print("Kurtosis:", kurtosis_value)
```

The mean and median of the sales data are both 165, indicating that the data is centered around 165. The dataset has multiple modes: 135, 145, 155, 165, and 175. The range is 90, showing the difference between the highest and lowest sales values. The standard deviation indicates the amount of variation in sales around the mean. The skewness is close to zero, suggesting that the dataset is approximately symmetric rather than positively or negatively skewed.
